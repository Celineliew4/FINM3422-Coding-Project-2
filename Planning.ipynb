### Step 2: Design Class Hierarchy (Object-Oriented Programming)

Start planning your Python classes using inheritance to avoid duplication.

- Base Class: Option
- Attributes: underlying, strike, expiry, option_type
- Methods: get_price(), get_delta()

Subclasses:
- EuropeanOption(Option)
- AmericanOption(Option)
- BarrierOption(EuropeanOption)
- BasketOption(Option)

Also consider a Portfolio class to hold all positions and compute aggregate hedge info.

### Step 3: Gather Market Data (as of 16 May 2025)

- Prices: Get closing prices for BHP, CBA, WES, CSL, WDS, MQG
- Volatility: Estimate implied or historical volatility
- Risk-Free Rates: Build a yield curve (e.g. from government bond yields)
- Dividends: Estimate future dividend yield for each stock

### Step 4: Implement Pricing Methods

Each derivative type may require a different method:

- European Options: Black-Scholes
- American Option: Binomial Tree or Finite Difference
- Barrier Option: Analytical (if assumptions allow) or Monte Carlo
- Basket Option: Monte Carlo simulation with correlated asset paths

Ensure you:

- Use correct interest rate discounting
- Handle dividend yields properly
- Validate each model with test cases


### Step 5: Implement Hedging Metrics

At a minimum:
- Delta for each option
- Aggregate portfolio delta
- High scoring:
    - Gamma, Vega (if possible)
    - Correlation-adjusted basket delta

### Step 6: Test Each Module

- Validate model outputs against known benchmarks (e.g. market prices or textbook cases)
